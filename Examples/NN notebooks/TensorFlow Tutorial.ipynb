TensorFlow is an end-to-end platform for machine learning. It supports the following:

- Multidimensional-array based numeric computation (similar to NumPy)

- Automatic differentiation

## Tensors

TensorFlow operates on multidimensional arrays or tensors represented as `tf.Tensor` objects. 

Tensors are multi-dimensional arrays with a uniform type (called a dtype). You can see all supported dtypes at `tf.dtypes.DType`

If you're familiar with NumPy, tensors are (kind of) like `np.arrays`

All tensors are immutable like Python numbers and strings: you can never update the contents of a tensor, only create a new one

Here is a two-dimensional tensor:

```python
#pip install tensorflow
```

```python
import tensorflow as tf

x = tf.constant([[1., 2., 3.],
                 [4., 5., 6.]])

print(x)
print(x.shape)
print(x.dtype)
```

The most important attributes of a tf.Tensor are its shape and dtype:

- Tensor.shape: tells you the size of the tensor along each of its axes.
- Tensor.dtype: tells you the type of all the elements in the tensor.

TensorFlow implements standard mathematical operations on tensors, as well as many operations specialized for machine learning.

For example:

```python
x + x
```

```python
5 * x
```

```python
# Note that the @ symbol is used for matrix multiplication
x @ tf.transpose(x)
```

```python
tf.concat([x, x, x], axis=0)
```

```python
# softmax converts all values to probabilities such that the sum is 1
# axis = -1 refers to the last axis
tf.nn.softmax(x, axis=-1)
```

```python
# Computes sum of all elements
tf.reduce_sum(x)
```

```python
c = tf.constant([[4.0, 5.0], [10.0, 1.0]])

# Find the largest value
print(tf.reduce_max(c))
# Find the index of the largest value
print(tf.math.argmax(c))
```

```python
tf.convert_to_tensor([1,2,3])
```

You can convert a tensor to a NumPy array using either `np.array` or the `tensor.numpy` method:

```python
import numpy as np
```

```python
np.array(x)
```

```python
x.numpy()
```

## Variables

Normal `tf.Tensor` objects are immutable. To store mutable elements in TensorFlow use `tf.Variable`

```python
var = tf.Variable([0.0, 0.0, 0.0])
```

```python
# Assign new elements to var
var.assign([1, 2, 3])
```

```python
# Update elements in var by adding 1 to each
var.assign_add([1, 1, 1])
```

## About shapes

Tensors have shapes. Some vocabulary:

- Shape: The length (number of elements) of each of the axes of a tensor
- Rank: Number of tensor axes. A scalar has rank 0, a vector has rank 1, a matrix is rank 2
- Axis or Dimension: A particular dimension of a tensor
- Size: The total number of items in the tensor, the product of the shape vector's elements

Tensors and `tf.TensorShape` objects have convenient properties for accessing these:

```python
rank_4_tensor = tf.zeros([3, 2, 4, 5])
print(rank_4_tensor.shape)
```

```python
print("Type of every element:", rank_4_tensor.dtype)
print("Number of axes:", rank_4_tensor.ndim)
print("Shape of tensor:", rank_4_tensor.shape)
print("Elements along axis 0 of tensor:", rank_4_tensor.shape[0])
print("Elements along the last axis of tensor:", rank_4_tensor.shape[-1])
print("Total number of elements (3*2*4*5): ", tf.size(rank_4_tensor).numpy())
```

## Indexing

```python
x_new = tf.transpose(x)
print(x_new)
```

Passing an integer for each index, the result is a scalar

```python
# Pull out a single value from a 2-D tensor
print(x_new[1, 1].numpy())
```

You can index using any combination of integers and slices:

```python
# Get row and column tensors
print("Second row:", x_new[1,:].numpy())
print("Second column:", x_new[:, 1].numpy())
print("Last row:", x_new[-1,:].numpy())
print("First item in last column:", x_new[0, -1].numpy())
print("Skip the first row:")
print(x_new[1:,:].numpy(), "\n")
```

You can also reshape a tensor:

```python
# You can reshape a tensor to a new shape
# Note that you're passing in a list
x = tf.constant([[1], [2], [3]])
reshaped = tf.reshape(x, [1, 3])
print(x.shape)
print(reshaped.shape)
```

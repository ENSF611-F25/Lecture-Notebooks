This tutorial uses the Titanic data set, stored as CSV. The data consists of the following data columns:

PassengerId: Id of every passenger.

Survived: Indication whether passenger survived. 0 for yes and 1 for no.

Pclass: One out of the 3 ticket classes: Class 1, Class 2 and Class 3.

Name: Name of passenger.

Sex: Gender of passenger.

Age: Age of passenger in years.

SibSp: Number of siblings or spouses aboard.

Parch: Number of parents or children aboard.

Ticket: Ticket number of passenger.

Fare: Indicating the fare.

Cabin: Cabin number of passenger.

Embarked: Port of embarkation.

```python
import pandas as pd
titanic = pd.read_csv("titanic.csv")
titanic.head()
```

How do I select a specific column of the dataframe?

I am interested in the ages of the passengers

```python
ages = titanic["Age"]
print(ages.head())
print(type(ages))
print(ages.shape)
```

How do I select multiple columns of the dataframe?

I am interested in the age and sex of the passengers

```python
age_sex = titanic[["Age", "Sex"]]
print(age_sex.head())
print(type(age_sex))
print(age_sex.shape)
```

How do I filter specific rows?

I am interested in passengers older than 35 years

```python
above_35 = titanic[titanic["Age"] > 35]
print(above_35.head())
print(above_35.shape)
```

I am interested in the average, minimum and maximum age of the passengers

```python
ages_min = titanic["Age"].min()
ages_max = titanic["Age"].max()
ages_avg = titanic["Age"].mean()
print(f"Min age: {ages_min}\nMax age: {ages_max}\nAvg age: {ages_avg:.2f}")
```

I am interested in passengers from cabin classes 2 and 3

Note: for multiple conditions, must use '&' or '|' instead of 'and' or 'or' because the base code for pandas is in C

```python
class_23 = titanic[titanic["Pclass"].isin([2, 3])]
print(class_23.head())
print(class_23.shape)
```

I want to work with passenger data for which the age is known

```python
age_no_na = titanic[titanic["Age"].notna()]
print(age_no_na.head())
print(age_no_na.shape)
```

How do I select a subset of columns and rows from the dataframe?

I'm interested in the names of the passengers older than 35 years

```python
adult_names = titanic.loc[titanic["Age"] > 35, "Name"]
adult_names.head()
```

I'm interested in rows 10 to 25 and columns 3 to 5

```python
subset = titanic.iloc[9:25, 2:5]
print(subset)
```
